### 1 - Clean the Prices

AmplifyMe quantify market making dataset assessment

Clean The Tape: Medium (difficulty level)

Desk context
You're on a market-making desk quoting three assets of varying liquidity, and everything starts from the prices tape, a per-timestamp reference price for Asset_1, Asset_2, and Asset_3. The raw feed is dirty: rows arrive out of order and occasionally duplicated, some prints are non-positive (a zero or a negative is never a real price), and there are gaps where nothing is printed. You can't quote off a dirty tape, so the first task is to repair it.

Problem
Implement clean_prices(prices) which returns a repaired copy of the tape, time-sorted and gap-filled.

Rules:

sort by timestamp and drop exact duplicate rows; reset the index to 0...n-1.
For each asset column, treat any value <= 0 as missing (NaN)
Forward-fill each asset column so gaps and stale bars carry the last good price. Never back-fill, because using a future price is look-ahead.
A leading gap (no earlier price exists) stays NaN.
Constraints

prices has a timestamp column plus Asset_1, Asset_2, Asset_3
Operate vectorized. pandas / numpy available
Sample I/O
Input (unsorted, one duplicate, one gap):
09:02, 100.2, 50.2, 20.2
09:00, 100.0, 50.0, 20.0
09:01, 100.1, NaN, 20.1
09:00, 100.0, 50.0, 20.0

Output:
09:00, 100.0, 50.0, 20.0
09:01, 100.1, 50.0, 20.1
09:02, 100.2, 50.2, 20.2

Notes
Foward-fill only. The leading NaN case is intentional: there is no price before the first print

#### Method 1

```python
import pandas as pd
import numpy as np

def clean_prices(prices: pd.DataFrame) -> pd.DataFrame:
### Repair the raw price tape.
### Args:
### prices: DataFrame with a 'timestamp' column and one column per asset (Asset_1, Asset_2, Asset_3). May contain duplicate rows, out-of-order timestamps,
### non-positive bad prints and NaNs.

### Returns:
### A cleaned DataFrame: time-sorted, de-duplicated, index reset, with non-positive prints treated as missing and gaps forward-filled.

    # Step 1: Sort, remove duplicates, reset index
    df = prices.sort_values('timestamp')
    df = df.drop_duplicates()
    df = df.reset_index(drop=True)

    # Step 2: change the <=0 values to NaN
    asset_cols = ['Asset_1', 'Asset_2', 'Asset_3']
    # where( ) keeps values where condition is True, replaces with NaN where False
    # so where(>0) keeps positive prices, turns zero/negative into NaN
    df[asset_cols] = df[asset_cols].where(df[asset_cols]>0)

    # Step 3: forward fill
    df[asset_cols] = df[asset_cols].ffill()

    return df
```

```python
data = {
    'timestamp': ['09:02', '09:00', '09:01', '09:00'],
    'Asset_1': [100.2, 100.0, 100.1, 100.0],
    'Asset_2': [50.2, 50.0, float('nan'), 50.0],
    'Asset_3': [20.2, 20.0, 20.1, 20.0]
}
prices = pd.DataFrame(data)
print(clean_prices(prices))
```

#### Method 2

```python
def clean_prices(prices: pd.DataFrame) -> pd.DataFrame:
    # Work on a copy so the original DataFrame is not modified.
    cleaned = prices.copy()

    # Put observations in chronological order, remove only exact duplicate
    # rows, and restore a consecutive index.
    cleaned = (
        cleaned
        .sort_values("timestamp", kind="stable")
        .drop_duplicates()
        .reset_index(drop=True)
    )

    # Every column other than timestamp is an asset-price column.
    asset_cols = [col for col in cleaned.columns if col != "timestamp"]

    # Non-positive prices are invalid, so convert them to missing values.
    cleaned[asset_cols] = cleaned[asset_cols].mask(
        cleaned[asset_cols] <= 0
    )

    # Carry the most recent valid price forward independently for each asset.
    # Leading NaNs remain NaN because no back-fill is performed.
    cleaned[asset_cols] = cleaned[asset_cols].ffill()

    return cleaned
```

```python
data = {
    'timestamp': ['09:02', '09:00', '09:01', '09:00'],
    'Asset_1': [100.2, 100.0, 100.1, 100.0],
    'Asset_2': [50.2, 50.0, float('nan'), 50.0],
    'Asset_3': [20.2, 20.0, 20.1, 20.0]
}
prices = pd.DataFrame(data)
print(clean_prices(prices))
```

### 2 - Debug the Tape Cleaner

The desk runs clean_tape over the raw market-data tape before any strategy reads it. It has been in production for months. Recently the traders have started reporting occasional bad prints slipping through stale fills appearing at the wrong time, the odd zero or negative price reaching the book, and a few rows that simply look wrong. Most of the time the output looks fine, which is why nobody caught it sooner. Your job is to find out what the cleaner is getting wrong and fix it.

The Job
clean_tape(prices) already ships(it is in the editor). It is supposed to satisfy the spec below exactly. It does not. Diagnose where it diverges and repair it so that it meets the contract for every input, not just the easy ones.

The Contract
Given a raw price tape, clean_tape must return a cleaned table that obeys all of these rules, in this order:
1. Sort the rows ascending by timestamp.
2. Drop duplicate timestamps, keeping the first occurrence after sorting.
3. Treat any price that is zero or negative as missing.
4. Forward-fill each asset column from its last valid (post-sort) price.
5. A leading missing value with no prior valid price stays null (it cannot be filled)

Forward-fill means: if a price is missing, carry forward the most recent valid price for that asset. Each asset column is filled independently.

Input Format
prices is a pd.DataFrame with these columns:

| Column | Type | Description |
|:-----|----:|:------:|
| timestamp | int | Milliseconds since session open |
| Asset_1 | float | Mid price for asset 1 (may be missing / 0 / negative on bad prints) |
| Asset_2 | float | Mid price for asset 2|
| Asset_3 | float | Mid price for asset 3|

Rows may arrive out of time order and the same timestamp may appear more than once.

Qutput Format

Return a pd.DataFrame with the same columns (timestamp, Asset_1, Asset_2, Asset_3):
1. One row per unique timestamp, sorted ascending by timestamp
2. Zero/negative/missing prints repaired by forward-fill per the contract
3. Index reset to 0, 1, 2, ...

Do not mutate the input DataFrame in place.


```python
import pandas as pd
import numpy as np

def clean_tape(prices):
    df = prices.copy()
    asset_cols = [c for c in df.columns if c != 'timestamp']

    # Step 1 (Sort): Put the tape into chronological order
    df = df.sort_values('timestamp').reset_index(drop=True)

    # Step 2 (Drop the Duplicates): Collapse repeated timestamps down to a single row
    # We should keep the FIRST oocurrence after sorting instead of LAST
    df = df.drop_duplicates(subset='timestamp', keep='first').reset_index(drop=True)

    # Step 3: Treat any price that is zero or negative as missing
    # where( ) keeps values where condition is True, replaces with NaN where False
    # so where(>0) keeps positive prices, turns zero/negative into NaN
    df[asset_cols] = df[asset_cols].where(df[asset_cols]>0)

    # Step 4 (Forward Fill): Carry the last good print forward over any gaps in the tape
    df[asset_cols] = df[asset_cols].ffill()

    # Step 5 (Leading NaN stays null)
    # ffill() only looks backwards, so if no prior valid price exists, NaN remains
    
    return df
    
```

```python
data = {
    'timestamp': [902, 900, 901, 900],
    'Asset_1':   [100.2, 100.0, 100.1, 100.0],
    'Asset_2':   [50.2, 50.0, float('nan'), 50.0],
    'Asset_3':   [20.2, 20.0, 20.1, 20.0]
}

prices = pd.DataFrame(data)
print("Input:")
print(prices)
print("\nOutput:")
print(clean_tape(prices))
```

### 3 - Measure Liquidity

Measure Liquidity
Desk context
Decisions such as, how wide to quote or how much to hold, depends on how liquid each asset is, and liquidity is not labelled in the data. You measure it from a clean, time-sorted prices tape and the requests flow: a liquid asset has a stable price and plenty of client interest; an illiquid one is jumpy and rarely asked for.
Problem
Implement liquidity_scores(prices, requests) which returns {asset:score} in [0.1, 1.0] (higher = more liquid), blending low return volatility with high request frequency.
Formula:
$$score = 0.1 + 0.9(0.5(1-\hat{vol})+0.5\hat{freq})$$
where vol is prices[asset].pct_change().std(), freq is the asset's share of all requests, and each is min-max normalised across the assets. The 0.1 + 0.9*(...) floor keeps even the least-liquid asset on a non-zero score, so it is still quotable (just wide) rather than frozen out.
Rules:
1.  Asset columns are every column of prices except timestamp.
2.  If a signal has zero range across assets, its normalised value is 0.5 for every asset.
3.  Round each score to 6 dp.
Constraints
1. At least 3 price rows. pandas/numpy available.
Sample I/O
Input: prices where Asset_1 is smooth, Asset .....
Output: {'Asset_1': 1.0, 'Asset_2': 0.511561...
Notes
1. Volatility is inverted (low vol = liquid);
frequency is kept (more flow = liquid). The [0.1, 1.0] floor is deliberate: the illiquid name keeps a small risk budget downstream instead of being rejected outright.

#### Method 1

```python
import pandas as pd
import numpy as np

def liquidity_scores(prices: pd.DataFrame, requests: pd.DataFrame) -> dict:

    ### Score each asset's liquidity in [0.1, 1.0] (higher = more liquid).

    ### Arg:
    ###     prices: clean, time-sorted DataFrame (timestamp + one col per asset).
    ###     requests: client RFQs (request_id, timestamp, asset, side, quantify).
    
    ### Returns:
    ###     {asset: score} rounded to 6 dp, blending low volatility and high
    ###     request frequency (each min-max normalised), floored onto [0.1, 1.0].

    # Step 1: Find all asset columns
    asset_cols = [c for c in prices.columns if c!= 'timestamp']

    # Step 2: Calculate vol for each asset
    # vol = std of % price changes; higher vol = less liquid
    vols = {}
    for asset in asset_cols:
        vols[asset] = prices[asset].pct_change().std()

    # Step 3: Calculate freq for each asset
    # freq = asset's share of all requests; higher freq = more liquid
    total_requests = len(requests)
    freqs = {}
    for asset in asset_cols:
        if total_requests == 0:
            freqs[asset] = 0 # no requests -> freq = 0
        else:
            freqs[asset] = (requests['asset'] == asset).sum() / total_requests

    # Step 4: Convert to arrays for min-max normalization
    vol_values = np.array(list(vols.values()))
    freq_values = np.array(list(freqs.values()))

    # Step 5: Min-max normalize vol
    # If range = 0 (all assets same vol), set all to 0.5 
    vol_range = vol_values.max() - vol_values.min()
    if vol_range == 0:
        hat_vol = np.full(len(asset_cols, 0.5))
    else:
        hat_vol = (vol_values - vol_values.min()) / vol_range

    # Step 6: Min-max normalize freq
    # If range = 0 (all assets same freq), set all to 0.5
    freq_range = freq_values.max() - freq_values.min()
    if freq_range == 0:
        hat_freq = np.full(len(asset_cols), 0.5)
    else:
        hat_freq = (freq_values - freq_values.min()) / freq_range

    # Step 7: Calculate scores and return dict
    scores = {}
    for i, asset in enumerate(asset_cols):
        raw = 0.1 + 0.9 * (0.5 * (1 - hat_vol[i]) + 0.5 * hat_freq[i])
        scores[asset] = round(float(raw), 6)

    return scores
```

```python
prices = pd.DataFrame({
    'timestamp': ['09:30', '09:31', '09:32', '09:33', '09:34'],
    'Asset_1': [100, 100.1, 100, 100.1, 100.05],
    'Asset_2': [50, 50.3, 49.8, 50.2, 49.9],
    'Asset_3': [20, 21, 19.5, 21.2, 19]
})

requests = pd.DataFrame({
    'request_id': [1, 2, 3, 4, 5],
    'timestamp': ['t', 't', 't', 't', 't'],
    'asset': ['Asset_1', 'Asset_1', 'Asset_1', 'Asset_2', 'Asset_3'],
    'side': ['BUY', 'SELL', 'BUY', 'BUY', 'BUY'],
    'quantity': [100, 100, 100, 100, 100]
})

print(liquidity_scores(prices, requests))
# Expected: {'Asset_1': 1, 'Asset_2': 0.511561, 'Asset_3': 0.1}
```

#### Method 2

```python
def liquidity_scores(
    prices: pd.DataFrame,
    requests: pd.DataFrame
) -> dict:
    asset_cols = [col for col in prices.columns if col != "timestamp"]

    # Raw volatility: standard deviation of percentage returns per asset.
    volatility = prices[asset_cols].pct_change().std()

    # Raw frequency: each asset's share of all client requests.
    request_counts = (
        requests["asset"]
        .value_counts()
        .reindex(asset_cols, fill_value=0)
    )

    if len(requests) == 0:
        frequency = pd.Series(0.0, index=asset_cols)
    else:
        frequency = request_counts / len(requests)

    def min_max_normalize(signal: pd.Series) -> pd.Series:
        minimum = signal.min()
        maximum = signal.max()

        # The specification assigns 0.5 when every asset has
        # the same value for this signal.
        if maximum == minimum:
            return pd.Series(0.5, index=signal.index, dtype=float)

        return (signal - minimum) / (maximum - minimum)

    normalized_volatility = min_max_normalize(volatility)
    normalized_frequency = min_max_normalize(frequency)

    # Lower volatility and higher request frequency mean more liquidity.
    blended = (
        0.5 * (1.0 - normalized_volatility)
        + 0.5 * normalized_frequency
    )

    scores = 0.1 + 0.9 * blended

    return {
        asset: round(float(scores[asset]), 6)
        for asset in asset_cols
    }
```

```python
prices = pd.DataFrame({
    'timestamp': ['09:30', '09:31', '09:32', '09:33', '09:34'],
    'Asset_1': [100, 100.1, 100, 100.1, 100.05],
    'Asset_2': [50, 50.3, 49.8, 50.2, 49.9],
    'Asset_3': [20, 21, 19.5, 21.2, 19]
})

requests = pd.DataFrame({
    'request_id': [1, 2, 3, 4, 5],
    'timestamp': ['t', 't', 't', 't', 't'],
    'asset': ['Asset_1', 'Asset_1', 'Asset_1', 'Asset_2', 'Asset_3'],
    'side': ['BUY', 'SELL', 'BUY', 'BUY', 'BUY'],
    'quantity': [100, 100, 100, 100, 100]
})

print(liquidity_scores(prices, requests))
# Expected: {'Asset_1': 1, 'Asset_2': 0.511561, 'Asset_3': 0.1}
```

### 4 - Fair Value At A Time

Fair Value at a Time
Desk context
Client requests arrive at arbitrary times that don't line up with your price dataset, so to quote you need the asset's reference price as of the request time. And you must refuse to quote when the most recent print is too old to trust, which may happen for an illiquid asset with all its gaps.
Problem
Implement reference_price(prices, asset, ts, max_staleness_minutes) which returns {'price': <float or None>}.
Rules:
1. price is the most recent valid (non-NaN, > 0) value of asset at or before ts.
2. If no valid price exists at or before ts, price is None.
3. If the last valid print is older than max_staleness_minutes before ts, price is None.
4. Do the lookup efficiently: a sorted as-of lookup, not a per-call scan.
Constraints
1. ts is a timestamp (string or pd.Timestamp). pandas available.
2. Always return a dict with the single key price.
Sample I/O
reference_price(prices, 'Asset_3', '09:32:30')
reference_price(prices, 'Asset_3', '09:50:00')
Notes
1. This is where the illiquid asset starts

```python
import pandas as pd
import numpy as np

def reference_price(prices: pd.DataFrame, asset: str, ts, max_staleness_minutes: float) -> dict:
    ### As-of reference price with a staleness cutoff.

    ### Args:
    ###    prices: clean DataFrame, 'timestamp' + one column per asset (may hold NaNs)
    ###    asset: which asset column to read.
    ###    ts: the request time.
    ###    max_staleness_minutes: refuse (price None) if the last valid print is older.

    ### Returns:
    ###    {'price': float} as of ts, or {'price': None} if missing / too stale.

    # Step 1: Convert the requested time ts into pandas Timestamp for time comparison
    query_ts = pd.Timestamp(ts)

    # Step 2: Filter rows with valid prices (non-NaN and positive)
    valid = prices[asset].notna() & (prices[asset] > 0)

    valid_times = pd.DatetimeIndex(
        pd.to_datetime(prices.loc[valid, "timestamp"])
    )
    valid_prices = prices.loc[valid, asset].to_numpy()

    # Step 3: If no valid prices exist at all, return None
    if len(valid_times) == 0:
        return {"price": None}

    # Step 4: Binary search for the last valid price at or before query_ts
    # searchsorted returns the insertion point; -1 gives the most recent valid price before ts 
    pos = valid_times.searchsorted(query_ts, side="right") - 1

    # Step 5: If all valid prices occur after ts, return None
    if pos < 0:
        return {"price": None}

    # Step 6: Check if the most recent valid price is too stale
    # If older than max_staleness_minutes before ts, refuse to quote
    last_time = valid_times[pos]
    if query_ts - last_time > pd.Timedelta(
        minutes=max_staleness_minutes
    ):
        return {"price": None}

    # Step 7: Return the valid reference price as a float
    return {"price": float(valid_prices[pos])}
```

```python
prices = pd.DataFrame({
    'timestamp': ['2026-06-15 09:30:00', '2026-06-15 09:31:00', 
                  '2026-06-15 09:32:00', '2026-06-15 09:33:00', 
                  '2026-06-15 09:34:00'],
    'Asset_3': [20, float('nan'), 20.2, float('nan'), float('nan')]
})

# Test Case 1
print(reference_price(prices, 'Asset_3', '2026-06-15 09:32:30', 10))
# Expected: {'price': 20.2}

# Test Case 2
print(reference_price(prices, 'Asset_3', '2026-06-15 09:50:00', 10))
# Expected: {'price': None}
```

### 5 - Debug the Fair-Value Lookup

Debug the fair-value lookup
A market-making desk values its book off an as-of price lookup given a timestamp, what was the fair value of an asset at that moment? The desk's fair_value_at helper is mispricing positions under some market conditions, and risk is flagging the discrepancies. The previous author has moved on. Your job: diagnose why it misprices and fix it so every condition is handled correctly.
The Contract
Implement fair_value_at(ts, asset, prices, max_staleness_minutes) so it returns the asset's fair value as of a query time, following the rules below exactly.
The code you start with already runs and returns plausible numbers, but it is subtly wrong. Read the spec, work out which conditions it mishandles, and correct it.
What "fair value as of ts" means
1. As-of (backward) lookup: return the price from the most recent bar at or before ts. You never look into the future.
2. Skip nulls: a bar can have a missing (null/NaN) price, skip null bars and keep walking back to the last bar that actually has a price.
3. Staleness clamp: if the most recent valid bar is older than max_staleness_minutes before ts, the quote is stale: return null. (Exactly max_staleness_minutes old is still
4. Before the first print: if ts is earlier than every bar in the series, there is nothing to look back to: return null.
5. Exact match: if a bar's timestamp equals ts exactly, that bar is used (when its price is valid).
The input series is not guaranteed to arrive in time order.
| Argument | Type | Description |
|:-----|----:|:------:|
| ts | str |  |
| asset | str |  |
| prices | pd.DataFrame | |
| max_staleness_minutes | int | |
Output Format
Return a dict:
{"price": <float>} # the fair value as of ...
{"price": None} # if no usable, fresh p...
Hints:
1. pd.to_datetime(...） parses both the column and the query time so you can compare them.
2. A duration in minutes: (a-b).total_seconds() / 60.0
3. Test your fix against unsorted input, missing prices, and a query that lands exactly on a bar.

```python
import pandas as pd
import numpy as np

def fair_value_at(ts, asset, prices, max_staleness_minutes):
    df = prices.copy()
    
    df['timestamp'] = pd.to_datetime(df['timestamp'])
    query_ts = pd.to_datetime(ts)

    # Bars strictly before the query time
    # Bug 1 fixed: <= instead of < (include exact match)
    # Bug 2 fixed: filter out NaN prices here directly
    eligible = df[
        (df['timestamp'] <= query_ts)
        & df[asset].notna()
        & (df[asset] > 0)
    ]
    
    if eligible.empty:
        return {'price': None}

    # Walk back to the most recent bar that actually has a price
    # Bug 3 fixed: sort first to handle unsorted input
    eligible = eligible.sort_values('timestamp')

    # Get the most recent valid bar
    chosen_idx = eligible["timestamp"].idxmax()
    chosen = eligible.loc[chosen_idx]

    # Bug 4 fixed: add staleness check
    age_minutes = (
        (query_ts - chosen["timestamp"]).total_seconds() / 60.0
    )

    if age_minutes > max_staleness_minutes:
        return {"price": None}

    return {"price": float(chosen[asset])}
```

```python
prices = pd.DataFrame({
    'timestamp': ['2024-01-15 09:00:00', '2024-01-15 09:05:00',
                  '2024-01-15 09:10:00', '2024-01-15 09:15:00',
                  '2024-01-15 09:20:00'],
    'Asset_X': [100, 101, 102, 103, 104]
})

# Test Case 1
print(fair_value_at('2024-01-15 09:23:00', 'Asset_X', prices, 10))
# Expected: {'price': 104}
```

### 6 - Spread From Liquidity

Spread from liquidity
Desk context
You can quote tight for a liquid asset and quote wide for an illiquid asset. You'are provided with the liquidity score and the volatility; turn them into a half-spread.
Problem
Implement half_spread(liquidity_score, volatility, params) where params has base, k_vol, k_liq and min_half_spread.
Formula:
half_spread = max(base + k_vol * vol + k_liq * (1 - liquidity_score), min_half_spread)
Constraints
liquidity_score in [0, 1], volatility >= 0.
Sample I/O
half_spread(1.0, 0.001, {'base': 0, 'k_liq': 0.5, 'k_vol': 5, 'min_half_spread': 0.01}
half_spread
Notes
Increases with volatility and as liquidity falls; clamped at the floor

```python
import pandas as pd
import numpy as np

def half_spread(liquidity_score: float, volatility: float, params: dict) -> float:
    ###
    ### Map liquidity + volatility to a half-spread.

    ### Args:
    ### liquidity_score: 0..1, higher = more liquid.
    ### volatility: short-term return volatility of the asset.
    ### params: {'base', 'k_vol', 'k_liq', 'min_half_spread'}

    ### Returns:
    ### Half-spread, never below params['min_half_spread'].

    spread = (params["base"] 
            + params["k_vol"] * volatility 
            + params["k_liq"]*(1.0 - liquidity_score))

    return max(spread, params["min_half_spread"])
```

```python
params = {'base': 0, 'k_liq': 0.5, 'k_vol': 5, 'min_half_spread': 0.01}

print(half_spread(1.0, 0.001, params))  # 流动性最好
print(half_spread(0.1, 0.05, params))   # 流动性最差
```

### 7 - Price One Request

Price One Request
Desk context
Quote a single client request around fair value, charging more for size and get the side exactly right. A client BUY means the market maker is SELLING, so you quote the ask (above reference); a client SELL means the maker is BUYING, so you quote the bid (below reference).
Problem
Implement quote_request(reference, half_spread, side, quantity, size_params, tick_size).
Formula:
BUY: [(ref + hs + kq) / tick] * tick
SELL: [(ref - hs - kq) / tick] * tick
with size penalty k = size_params['k'] and q = quantity.
Rules:
1. A bigger quantity worsens the price (the k*quantity penalty).
2. Round outward to the tick grid: ask up, bid down.
Constraints
side is 'BUY' or 'SELL'. numpy available.
Sample I/O
quote_request(100.0, 0.05, 'BUY', 100, {'k
quote_request(100.0, 0.05, 'SELL', 100, {'k
Notes
The side confusion is the trap: client BUY => maker sells the ask; client SELL => maker buys the bid.


```python
import pandas as pd
import numpy as np

def quote_request(reference, half_spread, side, quantity, size_params, tick_size) -> float:
    ###
    ### Price one client request.

    ### Args:
    ###    reference: fair value to quote around
    ###    half_spread: edge per side.
    ###    side: client's direction, 'BUY' (MM sells -> ask) or 'SELL' (MM buys -> bid).
    ###    quantity: order size (drives the size penalty).
    ###    size_params: {'k': size-penalty coefficient}.
    ###    tick_size: venue minimum increment.

    ### Returns:
    ###    The quoted price, snapped to the tick grid (ask up / bid down).

    # Extract the size penalty coefficient from params dict
    k = size_params["k"]

    # k*quantity penalty: a bigger quantity worsens the price
    size_penalty = k * quantity

    if side == "BUY":
        # Client BUYS -> market maker SELLS -> quote the ASK (above reference)
        raw_price = reference + half_spread + size_penalty

        # Round UP to the nearest tick so MM never quotes tighter than intended
        # np.ceil(raw / tick) * tick
        # Divide by tick, ceil/floor to nearest integer, multiply back to snap to tick grid
        quote = np.ceil(raw_price / tick_size) * tick_size

    else:
        # side == "SELL"
        # Client SELLS -> market maker BUYS -> quote the BID
        raw_price = reference - half_spread - size_penalty

        # Round DOWN to the nearest tick so MM never quotes tighter than intended
        # np.floor(raw / tick) * tick
        quote = np.floor(raw_price / tick_size) * tick_size

    return float(quote)
```

```python
# Test BUY
print(quote_request(100.0, 0.05, 'BUY', 100, {'k': 0.0001}, 0.01))
# raw = 100 + 0.05 + 0.0001*100 = 100.06
# ceil(100.06 / 0.01) * 0.01 = 100.06 ✅

# Test SELL
print(quote_request(100.0, 0.05, 'SELL', 100, {'k': 0.0001}, 0.01))
# raw = 100 - 0.05 - 0.0001*100 = 99.94
# floor(99.94 / 0.01) * 0.01 = 99.94 ✅
```

### 8 - Size And Impact

Size & Impact
Desk context
Filling a big order for an illiquid asset pushes the price, and the push accumulates across back-to-back requests; the thinner the asset, the harder it moves. Quote a sequence of requests in one asset, carrying that impact forward.
Problem
Implement
quote_stream_with_impact(reference, requests_for_asset, liquidity_score, impact_params) which returns the list of quoted prices (one per request, rounded to 9 dp). requests_for_asset is a list of {'side', 'quantity'}. Carry a running offset starting at 0.
Rules:
1. Quote off reference + offset: + half_spread for BUY, - half_spread for SELL
2. After each request: offset += k * signed_qty * (1 - liquidity_score + 1e-6), then offset *0 decay.
3. signed_qty is +quantity (BUY) or - quantity (SELL); impact_params has half_spread, k, optional decay.
Constraints
Order matters: each fill shifts the base for the next quote. numpy available.
Sample I/O
quote_stream_with_impact(20.0, [{'side': 'BUY {'half_spread': 0.1, -> [20.1, 20.145000045, 20.185500086]
Notes
A liquid asset (score ~1) barely moves; an illiquid one (~0) moves a lot

```python
import pandas as pd
import numpy as np

def quote_stream_with_impact(reference, requests_for_asset, liquidity_score, impact_params) -> list:
    ### 
    ### Quote a sequence of same-asset requests with cumulative, decaying impact.

    ### Args:
    ###    reference: starting fair value.
    ###    requests_for_asset: list of {'side': 'BUY' | 'SELL', 'quantity': float}
    ###    liquidity_score: 0..1; lower -> bigger price moves.
    ###    impact_params: {'half_spread', 'k', optional 'decay'}

    ### Returns:
    ###    List of quoted prices, one per request (rounded to 9 dp).


    # Extract params
    half_spread = impact_params["half_spread"]
    k = impact_params["k"]
    # Decay defaults to 1.0 (no decay) if not provided
    decay = impact_params.get("decay", 1.0)

    # Running price offset, starts at 0 before any fills
    offset = 0.0
    quotes = []

    for req in requests_for_asset:
        side = req["side"]
        quantity = req["quantity"]

        # Step 1: Quote using accumulated impact from prior fills
        # BUY:  client buys → MM sells → quote ASK (above reference)
        # SELL: client sells → MM buys → quote BID (below reference)
        if side == "BUY":
            quote = reference + offset + half_spread
            # BUY pushes price UP -> positive signed quantity
            signed_qty = quantity
        else: 
            quote = reference + offset - half_spread
            # SELL pushes price DOWN -> negative signed quantity
            signed_qty = -quantity

        quotes.append(round(float(quote), 9))

        # Step 2: Update offset based on this fill's market impact
        # Impact = k * signed_qty * (1 - liquidity_score + 1e-6)
        # - illiquid asset (score~0): (1-0+1e-6) ≈ 1.0 → large impact
        # - liquid asset   (score~1): (1-1+1e-6) ≈ 0.0 → tiny impact
        # 1e-6 ensures impact never fully vanishes even for perfectly liquid assets
        offset += (
            k
            * signed_qty
            * (1.0 - liquidity_score + 1e-6)
        )

        # Step 3: Apply decay to simulate market recovering between fills
        # e.g. decay=0.9 → each fill only carries 90% of its impact forward
        # decay=1.0 → impact fully persists (no recovery)
        offset *= decay

    return quotes
```

```python
# Test Case 1
reference = 20
liquidity_score = 0
requests = [
    {'side': 'BUY', 'quantity': 2500},
    {'side': 'BUY', 'quantity': 2500},
    {'side': 'BUY', 'quantity': 2500}
]
impact_params = {'k': 0.00002, 'decay': 0.9, 'half_spread': 0.1}

print(quote_stream_with_impact(reference, requests, liquidity_score, impact_params))
# Expected: [20.1, 20.145000045, 20.185500086]

# Test Case 2
reference = 100
liquidity_score = 1
print(quote_stream_with_impact(reference, requests, liquidity_score, impact_params))
# Expected: [100.1, 100.100000045, 100.100000085]
```

### 9 - Debug the Impact Model

```python
import math

def quote_impact_stream(reference, impact_params, liquidity_score, requests_for_asset) -> list:
    """
    Quote a sequence of same-asset requests with cumulative, decaying market impact.
    
    Each fill shifts the accumulated_impact state:
    - BUY orders push impact positive (price pressure up)
    - SELL orders push impact negative (price pressure down)
    - Impact mean-reverts by decay factor between requests
    """
    
    k = impact_params["k"]
    decay = impact_params["decay"]
    half_spread = impact_params["half_spread"]
    
    # Accumulated impact starts at 0 before any fills
    accumulated_impact = 0.0
    quotes = []
    
    for req in requests_for_asset:
        side = req["side"]
        quantity = req["quantity"]
        
        # BUY → ask above reference, SELL → bid below reference
        signed_spread = half_spread if side == "BUY" else -half_spread
        
        # Step 1: Quote using impact from PRIOR orders only
        # Bug 1 fixed: removed incorrect 'first' flag that double-counted
        # the spread (2x half_spread) on the very first quote
        quote = reference + signed_spread + accumulated_impact
        
        # Step 2: Record quote BEFORE updating impact
        # Bug 2 fixed: moved append here, BEFORE impact update
        # original code appended AFTER updating accumulated_impact,
        # which caused current order's impact to bleed into its own quote
        quotes.append(round(float(quote), 9))
        
        # Step 3: Calculate current order's signed market impact
        # Bug 3 fixed: removed incorrect + 1e-6 from impact formula
        # this question's spec uses (1 - liquidity_score) without 1e-6
        # unlike the previous question which explicitly required 1e-6
        side_sign = 1.0 if side == "BUY" else -1.0
        impact_term = k * quantity * (1.0 - liquidity_score)
        accumulated_impact += side_sign * impact_term
        
        # Step 4: Mean-revert the WHOLE accumulated state before next request
        # Bug 4 fixed: decay applied to WHOLE accumulated_impact, not just new term
        # original code: accumulated_impact += side_sign * (impact_term * decay)
        # which only decayed the new term, leaving old impact undecayed
        accumulated_impact *= decay
    
    return quotes
```

```python
import math

# Worked Example verification
reference = 200
impact_params = {'k': 0.01, 'decay': 0.5, 'half_spread': 0}
liquidity_score = 0
requests = [
    {'side': 'BUY', 'quantity': 100},
    {'side': 'BUY', 'quantity': 100},
    {'side': 'BUY', 'quantity': 100}
]
print(quote_impact_stream(reference, impact_params, liquidity_score, requests))
# Expected: [200.0, 200.5, 200.75]

# Test Case 1
reference = 100
impact_params = {'k': 0.001, 'decay': 0.5, 'half_spread': 0}
liquidity_score = 0
requests = [{'side': 'BUY', 'quantity': 500}]
print(quote_impact_stream(reference, impact_params, liquidity_score, requests))

# Test Case 2
reference = 250
impact_params = {'k': 0.002, 'decay': 0.7, 'half_spread': 0}
liquidity_score = 0.5
requests = [{'side': 'SELL', 'quantity': 300}]
print(quote_impact_stream(reference, impact_params, liquidity_score, requests))
# Expected: [250]
```

### 10 - Track the Book

```python
import pandas as pd
import numpy as np

def track_book(filled_requests: pd.DataFrame, prices: pd.DataFrame) -> dict:
    """
    Replay fills into per-asset inventory, cash, and mark-to-market PnL.
    
    The market maker always takes the OPPOSITE side of the client:
    - Client BUY  → maker SELLS → inventory decreases, cash increases
    - Client SELL → maker BUYS  → inventory increases, cash decreases
    
    Args:
        filled_requests: DataFrame with asset, side (client direction), quantity, fill_price.
        prices: clean DataFrame; last row is used to mark inventory.
    Returns:
        {'inventory': {asset: qty}, 'cash': float, 'pnl': float}
    """
    
    # Step 1: Identify all asset columns (everything except timestamp)
    # These become the keys of our inventory dict
    asset_cols = [c for c in prices.columns if c != "timestamp"]
    
    # Step 2: Start with a flat book - zero inventory and zero cash
    inventory = {asset: 0.0 for asset in asset_cols}
    cash = 0.0
    
    # Step 3: Replay every fill, updating inventory and cash
    # Key insight: maker takes OPPOSITE side of client
    for _, fill in filled_requests.iterrows():
        asset = fill["asset"]
        side = fill["side"]
        quantity = fill["quantity"]
        fill_price = fill["fill_price"]
        
        if side == "BUY":
            # Client BUY → maker SELLS
            # Maker loses inventory but gains cash
            inventory[asset] -= quantity
            cash += fill_price * quantity
        else:  # SELL
            # Client SELL → maker BUYS
            # Maker gains inventory but loses cash
            inventory[asset] += quantity
            cash -= fill_price * quantity
    
    # Step 4: Mark remaining inventory to market using last price row
    # Mark-to-market = what our inventory is worth at current prices
    last_prices = prices.iloc[-1]
    mark_to_market = sum(
        inventory[asset] * last_prices[asset]
        for asset in asset_cols
    )
    
    # Step 5: Calculate PnL = cash + mark-to-market value of inventory
    # The buckets must reconcile exactly: cash + mark_to_market = pnl
    pnl = cash + mark_to_market
    
    return {
        "inventory": inventory,
        "cash": round(float(cash), 6),
        "pnl": round(float(pnl), 6),
    }
```

```python
import pandas as pd
import numpy as np

# Test Case 1
prices = pd.DataFrame({
    'timestamp': ['09:30', '09:31'],
    'Asset_1': [100, 100],
    'Asset_2': [50, 50],
    'Asset_3': [20, 20]
})

filled_requests = pd.DataFrame({
    'asset': ['Asset_1', 'Asset_1'],
    'side': ['BUY', 'SELL'],
    'quantity': [100, 100],
    'fill_price': [100.05, 99.95]
})

print(track_book(filled_requests, prices))
# Expected: {'inventory': {'Asset_1': 0, 'Asset_2': 0, 'Asset_3': 0}, 'cash': 10.0, 'pnl': 10.0}



# Test Case 2
prices = pd.DataFrame({
    'timestamp': ['09:31'],
    'Asset_1': [100],
    'Asset_2': [50],
    'Asset_3': [20]
})

filled_requests = pd.DataFrame({
    'asset': ['Asset_1'],
    'side': ['SELL'],
    'quantity': [100],
    'fill_price': [99.95]
})

print(track_book(filled_requests, prices))
# Expected: {'pnl': 5, 'cash': -9995, 'inventory': {'Asset_1': 100, 'Asset_2': 0, 'Asset_3': 0}}
```

### 11 - Risk Limits And Skew

```python
import pandas as pd
import numpy as np

def decide_request(
    inventory: dict,
    request: dict,
    liquidity_score: float,
    params: dict
) -> dict:
    # --- Unpack inputs ---
    asset = request["asset"]
    side = request["side"]
    quantity = request["quantity"]
    reference = request["reference"]
    half_spread = request["half_spread"]

    base_limit = params["base_limit"]
    soft_frac = params["soft_frac"]
    skew_coeff = params["skew_coeff"]

    current_position = inventory.get(asset, 0)

    # --- Step 1: Compute hard and soft position limits ---
    # Less liquid assets (lower liquidity_score) get a smaller allowable
    # inventory limit.
    hard = base_limit * liquidity_score
    soft = soft_frac * hard

    # --- Step 2: Project the resulting inventory if this trade fills ---
    # Client BUY  => maker sells => maker's inventory decreases.
    # Client SELL => maker buys  => maker's inventory increases.
    signed = -quantity if side == "BUY" else quantity
    projected = current_position + signed

    # --- Step 3: Hard limit check -> reject outright ---
    # If the projected position would breach the hard cap, refuse the
    # trade entirely and don't produce a quote.
    if abs(projected) > hard:
        return {
            "action": "reject",
            "quote": None
        }

    # --- Step 4: Compute the inventory skew ---
    # Skew is based on CURRENT (pre-trade) inventory, not projected.
    # It nudges the quote to encourage trades that bring the position
    # back toward flat.
    skew = -skew_coeff * current_position

    # --- Step 5: Build the base quote around the skewed reference ---
    if side == "BUY":
        quote = reference + half_spread + skew
    else:  # SELL
        quote = reference - half_spread + skew

    # --- Step 6: Soft limit check -> widen instead of reject ---
    # Inside the hard limit but outside the soft limit: still quote,
    # but make the price less attractive by adding an extra half_spread
    # in the same direction as before.
    if abs(projected) > soft:
        action = "widen"
        if side == "BUY":
            quote += half_spread
        else:
            quote -= half_spread
    else:
        action = "accept"

    # --- Step 7: Return the decision, quote rounded to 6 dp ---
    return {
        "action": action,
        "quote": round(float(quote), 6)
    }
```

```python
# Test 1: Normal accept case - position within soft limit
inventory = {'Asset_1': 0}
request = {
    'asset': 'Asset_1',
    'side': 'BUY',
    'quantity': 10,
    'reference': 100.0,
    'half_spread': 0.5
}
liquidity_score = 1.0
params = {
    'base_limit': 100,
    'soft_frac': 0.5,
    'skew_coeff': 0.1
}
print("Test 1 (accept):", decide_request(inventory, request, liquidity_score, params))
# Expected: action=accept, quote = 100 + 0.5 + 0 = 100.5

# Test 2: Widen case - position outside soft but inside hard limit
inventory = {'Asset_1': -60}  # already short
request = {
    'asset': 'Asset_1',
    'side': 'BUY',
    'quantity': 10,
    'reference': 100.0,
    'half_spread': 0.5
}
print("Test 2 (widen):", decide_request(inventory, request, liquidity_score, params))
# projected = -60 - 10 = -70, soft=50, hard=100
# Expected: action=widen

# Test 3: Reject case - position would breach hard limit
inventory = {'Asset_1': -95}
request = {
    'asset': 'Asset_1',
    'side': 'BUY',
    'quantity': 10,
    'reference': 100.0,
    'half_spread': 0.5
}
print("Test 3 (reject):", decide_request(inventory, request, liquidity_score, params))
# projected = -95 - 10 = -105, hard=100
# Expected: action=reject, quote=None
```

### 12 - Debug the Risk-Limit Quoter

```python
import math

def price_quote(params, request, inventory, liquidity_score):
    soft_frac = params["soft_frac"]
    base_limit = params["base_limit"]
    skew_coeff = params["skew_coeff"]

    side = request["side"]
    asset = request["asset"]
    quantity = request["quantity"]
    reference = request["reference"]
    half_spread = request["half_spread"]

    current = inventory.get(asset, 0)

    # Bug 1 fixed: BUY should add to inventory, SELL should subtract
    signed_qty = quantity if side == "BUY" else -quantity
    projected = current + signed_qty

    hard_limit = base_limit
    soft_limit = soft_frac * hard_limit

    # Bug 2 fixed: >= instead of > (exactly at the cap must reject)
    if abs(projected) >= hard_limit:
        return {"quote": None, "action": "reject"}

    if abs(projected) > soft_limit:
        action = "widen"
        half = 2 * half_spread
    else:
        action = "accept"
        half = half_spread

    # Skew uses CURRENT pre-trade inventory
    skew = skew_coeff * current

    # Bug 3 fixed: subtract skew on both sides (long book -> quotes pull down)
    if side == "BUY":
        quote = reference + half - skew
    else:  # SELL
        quote = reference - half - skew

    return {"quote": round(float(quote), 4), "action": action}
```

```python
params = {"soft_frac": 0.5, "base_limit": 100, "skew_coeff": 0.1}

# Test 3: Accept - flat inventory, small BUY
inventory = {"Asset_1": 0}
request = {"side": "BUY", "asset": "Asset_1", "quantity": 10, "reference": 100.0, "half_spread": 0.5}
print("Test 3 (accept):", price_quote(params, request, inventory, 1.0))
# projected = 0+10=10, soft=50 -> accept
# Expected: quote = 100.5, action = accept

# Test 4: Widen - already long, BUY pushes further out (soft < proj < hard)
inventory = {"Asset_1": 55}
request = {"side": "BUY", "asset": "Asset_1", "quantity": 10, "reference": 100.0, "half_spread": 0.5}
print("Test 4 (widen):", price_quote(params, request, inventory, 1.0))
# projected = 65, soft=50, hard=100 -> widen
# skew = 0.1*55 = 5.5 (current, pre-trade)
# Expected: quote = 95.5, action = widen

# Test 5: Reject - projected exactly at hard limit
inventory = {"Asset_1": 90}
request = {"side": "BUY", "asset": "Asset_1", "quantity": 10, "reference": 100.0, "half_spread": 0.5}
print("Test 5 (reject, exact cap):", price_quote(params, request, inventory, 1.0))
# projected = 100 = hard_limit -> reject (>= boundary)
# Expected: quote = None, action = reject

# Test 6: Skew direction check - short inventory should pull quote UP
inventory = {"Asset_1": -55}
request = {"side": "SELL", "asset": "Asset_1", "quantity": 10, "reference": 100.0, "half_spread": 0.5}
print("Test 6 (short, SELL, widen):", price_quote(params, request, inventory, 1.0))
# projected = -65, soft=50, hard=100 -> widen
# skew = 0.1*(-55) = -5.5
# Expected: quote = 104.5, action = widen
```

### 13 - Hedge The Risk

```python
def hedge_inventory(
    prices: pd.DataFrame,
    inventory: dict
) -> dict:
    # Asset order must follow the price DataFrame.
    asset_cols = [c for c in prices.columns if c != "timestamp"]

    # Covariance matrix of percentage returns.
    returns = prices[asset_cols].pct_change()
    C = returns.cov().to_numpy()

    # Current inventory vector in the same asset order.
    v = np.array(
        [inventory.get(asset, 0.0) for asset in asset_cols],
        dtype=float
    )

    # The first two assets are the liquid hedge assets.
    liquid_assets = asset_cols[:2]

    # Covariance block for liquid assets and their covariance
    # against the whole portfolio.
    C_LL = C[:2, :2]
    C_L_all = C[:2, :]

    # Small ridge for numerical stability.
    C_LL_ridge = C_LL + 1e-12 * np.eye(2)

    # h_L = - C_LL^(-1) C_L,all v
    # solve() is numerically preferable to forming the inverse explicitly.
    hedge = -np.linalg.solve(
        C_LL_ridge,
        C_L_all @ v
    )

    # Portfolio after adding the hedge to the liquid positions.
    w = v.copy()
    w[:2] += hedge

    # Remaining portfolio variance.
    residual_variance = float(w.T @ C @ w)

    # Required rounding, while avoiding -0.0.
    hedge_result = {}
    for asset, qty in zip(liquid_assets, hedge):
        value = round(float(qty), 4)
        hedge_result[asset] = 0.0 if value == 0 else value

    residual_variance = round(residual_variance, 10)
    if residual_variance == 0:
        residual_variance = 0.0

    return {
        "hedge": hedge_result,
        "residual_variance": residual_variance
    }
```

```python
prices_dummy = pd.DataFrame({
    "timestamp": ["09:30", "09:31", "09:32"],
    "Asset_1": [100.0, 100.5, 100.7],
    "Asset_2": [50.0, 50.1, 50.4],
    "Asset_3": [20.0, 20.6, 20.1],
})
inventory_zero = {"Asset_1": 0, "Asset_2": 0, "Asset_3": 0}

result = hedge_inventory(prices_dummy, inventory_zero)
expected = {"hedge": {"Asset_1": 0.0, "Asset_2": 0.0}, "residual_variance": 0.0}

print("Result:  ", result)
print("Expected:", expected)
print("PASS" if result == expected else "FAIL")
```

### 14 - Debug the Hedge Optimizer

```python
def hedge_optimizer(prices: pd.DataFrame, inventory: dict) -> dict:
    """
    Minimum-variance hedge of an Asset_3 inventory using Asset_1 and Asset_2.
    """
    df = prices
    hedge_assets = ["Asset_1", "Asset_2"]
    target_asset = "Asset_3"

    # Only the Asset_3 exposure drives the hedge size.
    target_inventory = float(inventory.get(target_asset, 0.0))

    # Bug fixed: empty position -> nothing to hedge, return zeros directly.
    if target_inventory == 0.0:
        return {
            "hedge": {a: 0.0 for a in hedge_assets},
            "residual_variance": 0.0
        }

    # Step 1: Returns, not prices.
    # Convert each price column to simple percentage returns,
    # then drop the first row (undefined). Work entirely in return space.
    returns = df[hedge_assets + [target_asset]].pct_change(fill_method=None).iloc[1:]
    returns = returns.dropna()

    # Step 2: De-mean.
    # Centre the target returns and each instrument's returns by
    # subtracting their means.
    demeaned = returns - returns.mean()
    X = demeaned[hedge_assets].to_numpy()   # de-meaned Asset_1, Asset_2 returns
    y = demeaned[target_asset].to_numpy()   # de-meaned Asset_3 returns

    # Step 3: Solve the least-squares hedge.
    # OLS fit of de-meaned Asset_3 onto de-meaned [Asset_1, Asset_2].
    # No separate intercept term needed since everything is de-meaned.
    beta, _, _, _ = np.linalg.lstsq(X, y, rcond=None)

    # Step 4: Scale by inventory.
    # Hedge quantity for each instrument is -b_i * inventory["Asset_3"]
    # (you trade against your exposure).
    hedge_qty = -beta * target_inventory

    # Step 5: Residual variance.
    # Form the hedged return series: (target, de-meaned) - (instruments, de-meaned) @ beta.
    hedged_returns = y - X @ beta
    # Report the population variance: mean of the squared, de-meaned hedged returns.
    hedged_returns = hedged_returns - hedged_returns.mean()
    residual_variance = float(np.mean(hedged_returns ** 2))

    # Step 6: Rounding.
    # Round every hedge quantity and the residual variance to 4 decimal
    # places, and clean up any -0.0 artifacts.
    hedge_result = {}
    for asset, qty in zip(hedge_assets, hedge_qty):
        value = round(float(qty), 4)
        hedge_result[asset] = 0.0 if value == 0 else value

    residual_variance = round(residual_variance, 4)
    if residual_variance == 0:
        residual_variance = 0.0

    return {
        "hedge": hedge_result,
        "residual_variance": residual_variance
    }
```

```python
print("\n--- Exact-fit test (3 return rows, well-determined) ---")

# Returns designed so Asset_3 = 1*Asset_1 + 2*Asset_2 exactly, no noise.
r1 = (0.02, 0.01)   # Asset_1, Asset_2 returns, row 1
r2 = (-0.01, 0.02)  # row 2
r3 = (0.03, -0.01)  # row 3

a1 = [100.0]
a2 = [50.0]
a3 = [20.0]
for ra1, ra2 in [r1, r2, r3]:
    a1.append(a1[-1] * (1 + ra1))
    a2.append(a2[-1] * (1 + ra2))
    a3.append(a3[-1] * (1 + (ra1 + 2 * ra2)))

prices_exact = pd.DataFrame({
    "timestamp": [0, 1, 2, 3],
    "Asset_1": a1,
    "Asset_2": a2,
    "Asset_3": a3,
})

inventory_short = {"Asset_1": 0, "Asset_2": 0, "Asset_3": -100}
result = hedge_optimizer(prices_exact, inventory_short)
expected = {"hedge": {"Asset_1": 100.0, "Asset_2": 200.0}, "residual_variance": 0.0}

print("Result:  ", result)
print("Expected:", expected)
print("PASS" if result == expected else "FAIL")
```

### 15 - Client Toxicity And Markout

```python
def client_markout(
    fills: pd.DataFrame,
    prices: pd.DataFrame,
    horizon_minutes: float
) -> dict:
    fills = fills.copy()
    prices = prices.copy()

    # Step 1: Ensure proper datetime types and sort the price tape.
    fills["timestamp"] = pd.to_datetime(fills["timestamp"])
    prices["timestamp"] = pd.to_datetime(prices["timestamp"])
    prices = prices.sort_values("timestamp").reset_index(drop=True)

    horizon = pd.Timedelta(minutes=horizon_minutes)

    markouts = []
    for _, row in fills.iterrows():
        asset = row["asset"]
        target_time = row["timestamp"] + horizon

        # Step 2: As-of lookup - most recent price at or before target_time.
        # If horizon runs past the end of the tape, this naturally falls
        # back to the last available price (since all rows qualify).
        eligible = prices[prices["timestamp"] <= target_time]
        p_h = eligible.iloc[-1][asset] if not eligible.empty else prices.iloc[0][asset]

        fill_price = row["fill_price"]
        quantity = row["quantity"]

        # Step 3: Markout from the market maker's side.
        if row["side"] == "BUY":       # client bought -> MM sold
            markout = (fill_price - p_h) * quantity
        else:                          # client SELL -> MM bought
            markout = (p_h - fill_price) * quantity

        markouts.append({"client_id": row["client_id"], "markout": markout})

    # Step 4: Aggregate per client - average markout, round to 6 dp, toxic flag.
    markout_df = pd.DataFrame(markouts)
    result = {}
    for client_id, group in markout_df.groupby("client_id"):
        avg_markout = round(float(group["markout"].mean()), 6)
        result[client_id] = {
            "avg_markout": avg_markout,
            "toxic": avg_markout < 0
        }
    return result
```

```python
# ---- Test Case 1 ----
timestamps = pd.date_range("2026-06-15 09:30:00", periods=11, freq="1min")
prices = pd.DataFrame({
    "timestamp": timestamps,
    "Asset_1": [100.0 + 0.1 * i for i in range(11)],
    "Asset_2": [50.0 - 0.1 * i for i in range(11)],
})

fills = pd.DataFrame([
    {"client_id": "CLIENT_A", "timestamp": "2026-06-15 09:31:00", "asset": "Asset_1",
     "side": "BUY", "quantity": 100, "fill_price": 100.1},
    {"client_id": "CLIENT_B", "timestamp": "2026-06-15 09:31:00", "asset": "Asset_1",
     "side": "SELL", "quantity": 100, "fill_price": 100.1},
])

horizon_minutes = 5

result = client_markout(fills, prices, horizon_minutes)
expected = {
    "CLIENT_A": {"avg_markout": -50.0, "toxic": True},
    "CLIENT_B": {"avg_markout": 50.0, "toxic": False},
}

print("Result:  ", result)
print("Expected:", expected)
print("PASS" if result == expected else "FAIL")
```
